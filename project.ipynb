```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Smart Campus Canteen Management System

# Food items using NumPy array
food_items = np.array([
    "Burger", "Pizza", "Samosa",
    "Sandwich", "Dosa"
])

# Prepared food quantities
prepared = np.array([50, 40, 60, 35, 45])

# Sold food quantities
sold = np.array([42, 35, 55, 28, 38])

# Remaining / wasted food
wasted = prepared - sold

# Food prices
price = np.array([50, 80, 15, 40, 40])

# Calculate revenue using NumPy
revenue = sold * price

# Calculate waste percentage
waste_percentage = (wasted / prepared) * 100

# Create Pandas DataFrame
data = pd.DataFrame({
    "Food Item": food_items,
    "Prepared": prepared,
    "Sold": sold,
    "Wasted": wasted,
    "Price": price,
    "Revenue": revenue,
    "Waste Percentage": waste_percentage
})

while True:

    print("\n===== SMART CAMPUS CANTEEN =====")
    print("1. Display Food Details")
    print("2. Total Revenue")
    print("3. Best Selling Food")
    print("4. Food Waste Analysis")
    print("5. Average Sales")
    print("6. Food Demand Prediction")
    print("7. Display Graph")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":

        print(data.to_string(index=False))

    elif choice == "2":

        print("Total Revenue: Rs.", np.sum(revenue))

    elif choice == "3":

        index = np.argmax(sold)

        print("Best Selling Food:", food_items[index])
        print("Quantity Sold:", sold[index])

    elif choice == "4":

        print("Total Food Prepared:", np.sum(prepared))
        print("Total Food Sold:", np.sum(sold))
        print("Total Food Wasted:", np.sum(wasted))

        print("Waste Percentage:",
              round(np.sum(wasted) / np.sum(prepared) * 100, 2), "%")

    elif choice == "5":

        print("Average Sales:", np.mean(sold))

    elif choice == "6":

        print("\nNext Day Food Demand:")

        demand = np.ceil(sold * 1.10).astype(int)

        for i in range(len(food_items)):
            print(food_items[i], ":", demand[i], "units")

    elif choice == "7":

        plt.bar(food_items, sold)

        plt.title("Canteen Food Sales Analysis")
        plt.xlabel("Food Items")
        plt.ylabel("Quantity Sold")

        plt.show()

    elif choice == "8":

        print("Thank you for using the system!")
        break

    else:

        print("Invalid choice! Try again.")
```
